## A1. NumPy Arrays vs Python Lists
## Explanation
NumPy arrays are preferred for numerical work because they are more efficient for handling large amounts of numerical data. They usually use memory more efficiently than Python lists and allow mathematical operations to be performed on the whole array at once. NumPy also provides many built-in functions for calculations such as mean, sum, standard deviation, and matrix operations. Because of these features, NumPy is commonly used for data processing and numerical operations in AI and ML.

```python
import numpy as np

# Create a normal Python list
numbers_list = [10, 20, 30, 40, 50]

# Create the same data as a NumPy array
numbers_array = np.array([10, 20, 30, 40, 50])

# Multiplying a Python list by 2 repeats the list
list_result = numbers_list * 2

# Multiplying a NumPy array by 2 multiplies each element
array_result = numbers_array * 2

print("Python list result:", list_result)
print("NumPy array result:", array_result)
```

## A3. Element-wise Multiplication and Matrix Multiplication
## Explanation

`A * B` performs element-wise multiplication, where each value in A is multiplied by the corresponding value in B. `A @ B` performs matrix multiplication, where rows of A are multiplied with columns of B. In a neural network, element-wise multiplication can be used for applying a mask or scaling values. Matrix multiplication is commonly used to calculate the weighted inputs of a neural-network layer.


```python
import numpy as np

# Create two compatible 2D arrays
A = np.array([[1, 2],
              [3, 4]])

B = np.array([[5, 6],
              [7, 8]])

# Perform element-wise multiplication
element_wise = A * B

# Perform matrix multiplication
matrix_product = A @ B

print("A * B:")
print(element_wise)

print("A @ B:")
print(matrix_product)
```

## A4. Broadcasting in NumPy
## Explanation

Broadcasting allows NumPy to perform element-wise operations on arrays with different but compatible shapes. NumPy compares the dimensions of the arrays starting from the rightmost side. The dimensions are compatible when they are equal or when one of them is 1. If the dimensions do not follow these rules, NumPy gives a shape-related error.


```python
import numpy as np

# Create a 2D array
numbers = np.array([[1, 2, 3],
                    [4, 5, 6]])

# Create a single value
value = 10

# NumPy broadcasts 10 to every element of the array
result = numbers + value

print("Original array:")
print(numbers)

print("After broadcasting:")
print(result)
```

## A5. flatten() vs ravel()
## Explanation
`flatten()` converts an array into a one-dimensional array and always creates a copy of the data. `ravel()` also converts an array into one dimension, but it can return a view of the original array when possible. Since a view does not need to create another full copy, it can save memory. For a memory-constrained application working with large images, I would prefer `ravel()` because it can use less additional memory.


```python
import numpy as np

# Create a 2D array
image = np.array([[1, 2, 3],
                  [4, 5, 6]])

# Flatten creates a copy
flat_array = image.flatten()

# Ravel returns a flattened view when possible
ravel_array = image.ravel()

print("Original array:")
print(image)

print("Using flatten():")
print(flat_array)

print("Using ravel():")
print(ravel_array)
```

## A6. Understanding Axis 0 and Axis 1
## Explanation

In a 2D NumPy array, `axis=0` performs the operation down the rows, giving a result for each column. `axis=1` performs the operation across the columns, giving a result for each row. For example, `np.sum()` with `axis=0` calculates the sum of each column, while `axis=1` calculates the sum of each row.


```python
import numpy as np

# Create a 2D array
numbers = np.array([[1, 2, 3],
                    [4, 5, 6]])

# Sum of each column
column_sum = np.sum(numbers, axis=0)

# Sum of each row
row_sum = np.sum(numbers, axis=1)

print("Original array:")
print(numbers)

print("Sum along axis=0 (columns):")
print(column_sum)

print("Sum along axis=1 (rows):")
print(row_sum)
```

## A7. Normalization and Standardization
## Explanation

Normalization or standardization is important because different features may have different ranges or scales. If one feature has much larger values than another, it can affect the model's learning process. Normalization usually scales values to a fixed range such as 0 to 1. Standardization changes the data so that it has a mean of 0 and a standard deviation of 1.


```python
import numpy as np

# Create sample feature values
data = np.array([10, 20, 30, 40, 50])

# Min-Max normalization to the range 0 to 1
normalized = (data - np.min(data)) / (np.max(data) - np.min(data))

# Standardization using mean and standard deviation
standardized = (data - np.mean(data)) / np.std(data)

print("Original data:", data)
print("Normalized data:", normalized)
print("Standardized data:", standardized)
```

## C2. Z-Score Standardization
## Explanation
The function first calculates the mean and standard deviation using NumPy.
Each value is then converted into its z-score using `(x - mean) / std`.
The standardized values have a mean close to 0 and a standard deviation close to 1.
This makes the values easier to compare on the same scale.


```python
import numpy as np

# Define a function to standardize an array
def standardize(x):
    # Calculate the mean and standard deviation
    mean = np.mean(x)
    std = np.std(x)

    # Return the z-score standardized values
    return (x - mean) / std

# Create a sample array
data = np.array([10, 20, 30, 40, 50])

# Standardize the data
result = standardize(data)

# Display the result
print("Original data:", data)
print("Standardized data:", result)
```

## Explanation

## C5. Linear Model Prediction and MSE
## Explanation
The expression `X @ w + b` calculates the predicted values using matrix multiplication.
The predictions are compared with the actual values stored in `y_true`.
MSE calculates the average of the squared differences between actual and predicted values.
A smaller MSE means the predictions are closer to the actual values.


```python
import numpy as np

# Create input data with 3 samples and 2 features
X = np.array([[1, 2],
              [2, 3],
              [3, 4]])

# Create the weight vector
w = np.array([0.5, 1.0])

# Set the bias value
b = 2.0

# Create the actual target values
y_true = np.array([5, 7, 9])

# Calculate predictions using y_pred = X @ w + b
y_pred = X @ w + b

# Calculate Mean Squared Error
mse = np.mean((y_true - y_pred) ** 2)

print("Predictions:", y_pred)
print("MSE:", mse)
```

## D1. Data Preprocessing Pipeline

### Step 1: Detect Missing Values

We can use `np.isnan()` to identify the `NaN` values in the array.
`np.any(np.isnan(data))` checks whether the array contains at least one missing value.
This helps us identify missing data before processing it.

### Step 2: Handle Missing Values

For this example, I will replace the missing values with the mean of their respective columns.
Mean imputation is useful when only a small number of values are missing and the remaining values are available.
After replacing the `NaN` values, the data can be used for further processing.

### Step 3: Standardize the Data

After handling the missing values, we calculate the mean and standard deviation of each column.
Using `(data - mean) / std`, each feature is standardized separately.
This gives the features a mean close to 0 and a standard deviation close to 1.

## Explanation 
The pipeline detects missing values, replaces them using column means, and then standardizes each feature so the cleaned data is ready for a machine-learning model.


```python
import numpy as np

# Create a sample 2D array containing a NaN value
data = np.array([[10, 20],
                 [15, np.nan],
                 [20, 40],
                 [25, 50]], dtype=float)

# Step 1: Detect missing values
missing_values = np.isnan(data)

print("Missing value locations:")
print(missing_values)

print("Contains NaN:", np.any(missing_values))

# Step 2: Replace NaN values with the column mean
column_means = np.nanmean(data, axis=0)

# Find the positions of NaN values
rows, columns = np.where(np.isnan(data))

# Replace each NaN with the mean of its column
data[rows, columns] = column_means[columns]

print("\nData after handling NaN:")
print(data)

# Step 3: Standardize the clean data
mean = np.mean(data, axis=0)
std = np.std(data, axis=0)

standardized_data = (data - mean) / std

print("\nStandardized data:")
print(standardized_data)
```

## E1. Per-Column Min-Max Normalization

Min-max normalization can be applied separately to each feature column.
Using `axis=0`, NumPy calculates the minimum and maximum value of each column.
The formula `(X - min) / (max - min)` scales every column independently to the range 0 to 1.
This is fully vectorized and does not require any explicit Python loops

## Explanation
Each feature column is normalized independently, so the minimum becomes `0` and the maximum becomes `1` in every column.


```python
import numpy as np

# Create a 2D feature matrix
X = np.array([[10, 100, 1000],
              [20, 200, 2000],
              [30, 300, 3000],
              [40, 400, 4000]], dtype=float)

# Find the minimum value of each column
col_min = np.min(X, axis=0)

# Find the maximum value of each column
col_max = np.max(X, axis=0)

# Apply min-max normalization column-wise
X_normalized = (X - col_min) / (col_max - col_min)

# Display the results
print("Original matrix:")
print(X)

print("\nColumn minimums:")
print(col_min)

print("\nColumn maximums:")
print(col_max)

print("\nNormalized matrix:")
print(X_normalized)
```

## E2. Three-Level Score Classification

The scores can be classified using nested `np.where()` without using `if/else` statements or loops.
Scores below 40 are labeled **Fail**.
Scores from 40 to 70 are labeled **Average**, and scores above 70 are labeled **Excellent**.
The entire classification is performed as one vectorized expression.
## Explanation
 The first `np.where()` checks whether a score is below 40.
If it is not, the second `np.where()` checks whether it is 70 or below.
